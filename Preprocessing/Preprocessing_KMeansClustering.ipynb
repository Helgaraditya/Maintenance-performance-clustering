## Load Data

```python
df = pd.read_excel('data order ( Dimension_fact).xlsx')
df.head()
```

## Merge Data

```python
df_merged = pd.merge(df_droprows, df2, on="Order", how="inner")
```

```python
df_merged.head()
```

```python
# df_merged.isnull().sum()
```

## Mengambil Kolom yang diperlukan

```python
data = df_merged[[
    "Order", "Description", "Planner group", "Description2", "Actual start", "Actual Finish",
    "Nama Sparepart","Base Unit", "Quantity", "Amount in LC"
]]
```

```python
data = data.rename(columns={'Description':'Maintenance','Description2':'Equipment Dikerjakan'})
```

```python
data.isnull().sum()
```

# Membuat kolom baru untuk durasi pengerjaan

```python
data['Duration'] = (data['Actual Finish'] - data['Actual start']).dt.days
```

## Pengelompokan dengan Fungsi Aggregasi 

```python
grouped_order = data.groupby("Planner group").agg(
    Jumlah_maintenance=("Maintenance", "count"),
    total_amount=("Amount in LC", "sum"),
    average_duration=("Duration", "mean")
).reset_index()
```

## Normalisasi dengan MinMaxScaler

```python
x = grouped_order.iloc[:,[1,2,3]]

y = (x-x.min())/(x.max()-x.min())
y
```

```python
# y.reset_index(drop=True)
```

## Kmeans Clustering

```python
import warnings
```

```python
wcs=[]
for i in range (1,11):
  kmeans=KMeans(n_clusters=i, init='k-means++', random_state=39)
  kmeans.fit(y)
  wcs.append(kmeans.inertia_)
plt.plot(range(1,11),wcs)
plt.title('Elbow Method')
plt.xlabel('k')
plt.ylabel('Withn Cluster Sum of Square')
#plt.savefig("elbow.pdf", format="pdf", bbox_inches="tight")
plt.show()
```

```python
from sklearn.metrics import silhouette_score
no_of_clusters=[2,3,4,5,6]
for n_clusters in no_of_clusters:
   cluster = KMeans(n_clusters = n_clusters)
   cluster_labels = cluster.fit_predict(y)
   silhouette_avg = silhouette_score(y, cluster_labels)
   print('For no of clusters =', n_clusters,'The average silhouette_score is :',silhouette_avg)
```

```python
kmeans=KMeans(n_clusters=2,init='k-means++', max_iter=300,n_init=10,random_state=101)
z_kmeans=kmeans.fit_predict(y)

grouped_order['cluster']=z_kmeans+1
grouped_order
```

```python
import os
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Set OMP_NUM_THREADS to 1 to avoid memory leak warning
os.environ['OMP_NUM_THREADS'] = '1'

# Example data
X = [[1, 2], [1, 4], [1, 0],
     [4, 2], [4, 4], [4, 0]]

# Initialize KMeans with explicit n_init
n_samples = len(X)

for n_clusters in range(2, n_samples):
    kmeans = KMeans(n_clusters=n_clusters, n_init=10)
    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)
    print(f"For no of clusters = {n_clusters} The average silhouette_score is : {score}")
```
